# Variáveis aleatórias - Aula 4

Esse código faz uma simulação de um jogo de dados para estimar a probabilidade de obter lucro em uma aposta de R$ 5.

```python
# bibliotecas
from random import randint

aposta = 5 # valor da aposta
Ns = 1000 # numero de simulações/jogos
p = 0 # probabilidade de ganhar

# simulação
for n in range(1, Ns):
    G = 0 # lucro do jogador
    y = randint(1, 6) # sorteia um número de 1 a 6
    x = randint(1, 6) # sorteia outro número de 1 a 6
    if y > x:  # se o valor do jogador é maior
        G = 2 * (y - x) # lucro do jogador

        if G > aposta:  # se houver lucro
            p = p + 1 # incrementa a contagem de lucros

p = p/Ns # probabilidade de ganhar

print("Aposta: ", aposta, "reais.")
print("Valor teórico:", format(1/6, '.2f'))
print("Chance de lucro: ", format(p, '.2f'))
```
