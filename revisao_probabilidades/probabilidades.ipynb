# Probabilidades - Aula 1

Simulação do lançamento de uma moeda 10 vezes, considerando uma probabilidade de 60% de obter cara (p = 0.6). 

Em cada lançamento, um número aleatório entre 0 e 1 é gerado. Se esse número for menor que 0,6, o resultado é registrado como cara ('H'); caso contrário, é registrado como coroa ('T'). 

Ao final da simulação, o programa exibe a sequência de resultados obtidos, o número de caras, o número de coroas e a proporção de caras observada nas simulações.

```python
# Bibliotecas
import numpy as np

p = 0.6 # probabilidade de obter uma cara
nsim = 10 # número de simulações
nhead = 0 # número de caras
saida = [] # saída das simulações

# Simulação de lançamentos de moeda
for i in range(0, nsim):
    if np.random.rand() < p:
        nhead += 1
        saida.append('H')
    else:
        saida.append('T')

print(f'Saída: {saida}') # Imprime a saída das simulações
print(f'Número de caras: {nhead}') # Imprime o número de caras
print(f'Número de coroas: {nsim - nhead}') # Imprime o número de coroas
print(f'Proporção de caras: {nhead/nsim:.2f}') # Imprime a proporção de caras
```

Simulação para ilustrar a **Lei dos Grandes Números**. 
Para diferentes quantidades de lançamentos de uma moeda (de 1 até 1.000, em incrementos de 10), são simulados lançamentos com probabilidade de 60% de obter cara (p = 0.6). 

Em cada conjunto de simulações, é calculada a proporção de caras obtidas e esse valor é armazenado. 

Ao final, é gerado um gráfico que mostra como a proporção observada de caras tende a se aproximar da probabilidade teórica (0,6), representada por uma linha tracejada vermelha, à medida que o número de simulações aumenta.

*Obs. Testar com diferentes números de simulações para verificar a convergência!*

```python
# Bibliotecas
import numpy as np
import matplotlib.pyplot as plt

p = 0.6 # probabilidade de obter uma cara
vp = [] # vetor para armazenar a proporção de caras
vsim = [] # vetor para armazenar o número de simulações
Nsim = 1000 # número máximo de simulações

# Simulação de lançamentos de moeda
for nsim in np.arange(1, Nsim, 10):
    nhead = 0
    for i in range(1, nsim):
        if np.random.rand() < p:
            nhead += 1

    vp.append(nhead/nsim)
    vsim.append(nsim)   

# Gráfico da convergência da proporção de caras
plt.figure(figsize=(10, 6))
plt.plot(vsim, vp, 'o', markersize=2)
plt.axhline(y=p, color='r', linestyle='--', label=f'Probabilidade real (p={p})')
plt.xlabel('Número de Simulações')
plt.ylabel('Proporção de Caras')
plt.xlim(0, Nsim)
plt.ylim(0, 1)
plt.title('Convergência da Proporção de Caras')
plt.grid(True)
plt.show()
```
