## 2. Implemente uma função que calcule o fatorial de um número.

```python
# Calcula o fatorial de um número
def calcular_fatorial(n):
    if n < 0:
        return "Fatorial não definido para números negativos."
    elif n == 0 or n == 1:
        return 1
    else:
        fatorial = 1
        for i in range(2, n + 1):
            fatorial *= i
        return fatorial

# Solicita ao usuário que digite um número para calcular o fatorial
numero = int(input("Digite um número para calcular o fatorial: "))
print("O número digitado foi:", numero)
print("O fatorial de", numero, "é:", calcular_fatorial(numero))
```

## 3. Gere os primeiros termos da sequência de Fibonacci.

```python
'''
A sequência de Fibonacci é uma série numérica onde cada número é a soma dos dois números anteriores, começando por 0 e 1. 
A fórmula matemática para calcular o n-ésimo termo da sequência de Fibonacci é: 
F(n) = F(n-1) + F(n-2), onde F(0) = 0 e F(1) = 1
'''
# Função para calcular a sequência de Fibonacci
def fibonacci(n):
    a, b = 0, 1

    for _ in range(n): # Itera n vezes para gerar os termos da sequência de Fibonacci
        print(a, end=" ") # Exibe o termo atual da sequência de Fibonacci
        a, b = b, a + b # Atualiza os valores de a e b para os próximos termos da sequência

# Exemplo de uso
quantidade = int(input("Digite a quantidade de termos: "))
print("A sequência de Fibonacci com", quantidade, "termos é:")
fibonacci(quantidade)  
```

## 4. Escreva um programa que encontre os números primos em um intervalo definido pelo usuário.


```python
'''
Números primos são números naturais maiores que 1 que possuem apenas dois divisores: 1 e ele mesmo.
    1. O usuário informa o início e o fim do intervalo.
    2. O programa percorre todos os números desse intervalo.
    3. Para cada número maior que 1, verifica se ele possui divisores além de 1 e dele mesmo.
    4. Se não encontrar nenhum divisor, o número é considerado primo e é exibido.
'''
inicio = int(input("Digite o início do intervalo: "))
fim = int(input("Digite o fim do intervalo: "))

print("Números primos no intervalo:")

for num in range(inicio, fim + 1):
    if num > 1:
        primo = True

        for i in range(2, int(num ** 0.5) + 1):
            if num % i == 0:
                primo = False
                break

        if primo:
            print(num)
```

## 6. Gere dois vetores aleatórios independentes e represente os resultados em um gráfico de dispersão.

```python
import numpy as np
import matplotlib.pyplot as plt

x  = np.random.rand(50)  # Gera um array de 5 números aleatórios entre 0 e 1
y = np.random.rand(50)  # Gera outro array de 5 números aleatórios entre 0 e 1

plt.figure(figsize=(8, 6))  # Define o tamanho da figura
plt.scatter(x, y, color='blue')  # Cria um gráfico de dispersão com os pontos (x, y)
plt.title("Gráfico de Dispersão")  # Adiciona um título ao gráfico
plt.xlabel("Eixo X")  # Adiciona um rótulo ao eixo X
plt.ylabel("Eixo Y")  # Adiciona um rótulo ao eixo Y    
plt.show()  # Exibe o gráfico
```

## 7. Gere dados com distribuição Gaussiana e represente o respectivo histograma.

## 8. Gere uma tabela em que as colunas seguem distribuições exponenciais com diferentes taxas e salve os valores em um arquivo.

```python
# bibliotecas
import numpy as np
import pandas as pd

np.random.seed(42)

# Número de amostras
n = 100

# Gera colunas com diferentes taxas (lambda)
dados = pd.DataFrame({
    "lambda_0.5": np.random.exponential(scale=1/0.5, size=n),
    "lambda_1.0": np.random.exponential(scale=1/1.0, size=n),
    "lambda_2.0": np.random.exponential(scale=1/2.0, size=n),
    "lambda_5.0": np.random.exponential(scale=1/5.0, size=n)
})

# Salva em um arquivo CSV
dados.to_csv("new_data/distribuicao_exponencial.csv", index=False)

print(70 * "-")
print("Arquivo 'distribuicao_exponencial.csv' salvo com sucesso!")
print(70 * "-")
print(dados.head())
```

## 9. Crie um dicionário com nomes de alunos e notas. Depois, calcule a média da turma.

## 10. Crie uma função que receba um vetor NumPy e retorne sua média, seu desvio padrão e seu valor máximo

```python
import numpy as np

def estatisticas_basicas(notas):
    media = np.mean(notas) # Calcula a média das notas
    desvio_padrao = np.std(notas) # Calcula o desvio padrão das notas
    maximo = np.max(notas) # Calcula o valor máximo das notas
    minimo = np.min(notas) # Calcula o valor mínimo das notas
    return media, desvio_padrao, maximo, minimo

estatisticas = estatisticas_basicas(notas) # Chama a função estatisticas_basicas() passando a lista de notas como argumento
print(70 * "-")
print("Média:", estatisticas[0]) # Exibe a média das notas
print("Desvio Padrão:", estatisticas[1]) # Exibe o desvio padrão das notas
print("Máximo:", estatisticas[2]) # Exibe o valor máximo das notas          
print("Mínimo:", estatisticas[3]) # Exibe o valor mínimo das notas          
print(70 * "-")
```
